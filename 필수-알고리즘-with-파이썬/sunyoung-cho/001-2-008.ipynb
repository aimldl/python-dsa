```
# 001: 반복문을 사용하여 0부터 n까지의 합 출력하기
def print_to_n(n):
    i = 0
    sum = 0
    while (i<=n):
      sum = sum + i
      i = i + 1
    print(sum)
```

```
print_to_n(5)
```

```
print_to_n(100)
```

```
# 002: 재귀 호출을 사용하여 1부터 20까지 출력하기

def print_to_n2(n):
    if n >0:
       print_to_n2(n-1)
    print(n)
```

```
# 007: 369 게임만들기

# 3의 배수가 아닌 경우는 숫자를 출력하고, 3의 배수인 경우는 "X"를 출력하게 함

def solve_369(n):
    i = 1
    while i <= n:
        if i%3==0:
           print("X")
        else:
           print(i)
        i= i+1
```

```
solve_369(10)
```

```
# 008: 자연수 n이 소수인지 아닌지를 출력하기

# 주어진 숫자가 소수(prime number)라는 것은 0보다 큰 자연수인 약수가 1과 자기 자신뿐인 1보다 큰 자영수를 말함
# 사용자가 입력한 숫자 n이 소수인지 아닌지를 확인하는 코드 만들기

def check_prime(n):
   i=2
   while i < n:
      if n % i == 0:
        break
      i= i+1
   if i == n:
      print(n, " 은 소수")
   else:
      print(n, " 은 합성수")
```

```
check_prime(19)
```

```
check_prime(4)
```
