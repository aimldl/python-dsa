```
# 009: 2~N 사이의 모든 소수를 추출하기
def check_prime(n):
   i = 2
   while i < n:
     if n % i == 0:
       break
     i = i + 1

   if i == n:
       print(n, "은 소수임")
   else:
       print(n, "은 합성수임")
       
```

```
check_prime(5)
```

```
i = 2
while i<=20:
    check_prime(i)
    i = i+1
```

```
# N을 넣으면 2~N 사이의 소수만 출력하는 프로그램으로 바꾼다면

def output_only_prime_below(N):
    j = 2
    while j<=N:
       check_prime(j)
       j = j+1
```

```
check_prime(2)
```

```
output_only_prime_below(5)
```

```
# 소수인 경우만 출력
def output_when_prime(n):
   i = 2
   while i < n:
     if n % i == 0:
       break
     i = i + 1

   if i == n:
       print(n, "은 소수임")
```

```
# N을 넣으면 2~N 사이의 소수만 출력하는 프로그램으로 바꾼다면

def output_only_prime_below(N):
    j = 2
    while j<N:
       output_when_prime(j)
       j = j+1
```

```
output_only_prime_below(5)
```
