## 01\. 반복문을 사용하여 0부터 n까지의 합 출력하기, pp.152-154
### Problem
* 0부터 n까지 연속된 숫자의 합을 계산하여 화면에 출력하는 프로그램을 작성해봅시다.
* 단, n은 사용자가 입력해도 되고 직접 코드에 작성해도 무관합니다.

### Hint

### Solution

#### while 루프를 사용한 경우

````
def print_sum_upto(n):
    sum = 0
    i = 0
    while i <= n:
        sum += i
        print(sum)
        i += 1
````

````
# Unit test
print_sum_upto(20)
````

#### for 루프를 사용한 경우

````
def print_sum_upto(n):
    sum = 0
    for i in range(0, n + 1):
        sum += i
        print(sum)
````

````
# Unit test
print_sum_upto(20)
````

#### 순환 (recursion)을 사용한 경우

````
# 아래 수식 때문에 함수명을 print_sum_upto가 아닌 sum_upto로 했습니다.
#   sum = n + sum_upto(n-1)
def sum_upto(n):
    if n == 0:
        return 0
    
    sum = n + sum_upto(n-1)
    print(sum)
    return sum
````

````
# Unit test
sum_upto(20)
````
